Age: Idade do paciente em anos (valor numérico contínuo).

Sex: Gênero do paciente 0: Feminino (female) 1: Masculino (male).

Chest pain type: Tipo de dor no peito (categórico: 1 = angina típica, 2 = angina atípica, 3 = dor não anginosa, 4 = assintomática).

BP: Pressão arterial em repouso em mmHg (valor numérico contínuo).

Cholesterol: Nível de colesterol sérico em mg/dl (valor numérico contínuo).

FBS over 120: Açúcar no sangue em jejum > 120 mg/dl (binário: 1 = sim, 0 = não).

EKG results: Resultados do eletrocardiograma em repouso (categórico: 0 = normal, 1 = anormalidade da onda ST-T, 2 = hipertrofia ventricular esquerda provável ou definitiva).

Max HR: Frequência cardíaca máxima alcançada durante o exercício (valor numérico contínuo, em bpm).

Exercise angina: Angina induzida por exercício (binário: 1 = sim, 0 = não).


ST depression: Depressão do segmento ST induzida pelo exercício em relação ao repouso (valor numérico contínuo, em mm).

Slope of ST: Inclinação do segmento ST no pico do exercício (categórico: 1 = ascendente, 2 = plana, 3 = descendente).

Number of vessels fluro: Número de vasos principais coloridos por fluoroscopia (categórico: 0-3, indicando obstrução coronariana).

Thallium: Resultado do teste de estresse com tálio (categórico: 3 = normal, 6 = defeito fixo, 7 = defeito reversível).
